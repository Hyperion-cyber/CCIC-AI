Which is more selective: the University of Colorado Boulder or Colorado State University-Fort Collins?
This is an observational study.

```python
import pandas as pd
import seaborn as sns
import matplotlib as plt
college = pd.read_csv("cleaned_college.csv")
college
```

```python
schools = ["University of Colorado Boulder",
           "Colorado State University-Fort Collins"]

admission_rate = college[college["school_name"].isin(schools)][["school_name", "admission_rate"]]
print(admission_rate)
```

```python
data = college[college["school_name"].isin(schools)]

sns.barplot(data=data, x="school_name", y="admission_rate")
```

CU Boulder is more selective, at about 78.1% vs. 88.5%.
