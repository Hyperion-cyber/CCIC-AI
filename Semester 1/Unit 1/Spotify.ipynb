What was the average bpm of popular 2023 songs?

This is a ... study.

```python
import pandas as pd
import seaborn as sns
```

```python
spotify = pd.read_csv("spotify-2023.csv", encoding="latin-1")
spotify
```

```python
avg_bpm = spotify["bpm"].mean()
print(avg_bpm)
```

```python
mean_bpm = spotify['bpm'].mean()
median_bpm = spotify['bpm'].median()

sns.set_theme(style='whitegrid')
ax = sns.histplot(spotify['bpm'], bins=10, kde=True, color='steelblue')
ax.axvline(mean_bpm, color='red', linestyle='--', label=f'Mean: {mean_bpm:.1f}')
ax.axvline(median_bpm, color='green', linestyle='--', label=f'Median: {median_bpm:.1f}')
ax.set_title('Distribution of BPM (Spotify 2023)')
ax.set_xlabel('BPM')
ax.set_ylabel('Count')
ax.legend()
ax.figure.tight_layout()
ax.figure.savefig('bpm_histogram.png', dpi=150)
```

The average bpm is 122.5.
