How do the steak preferences of men differ from the preferences of women?
This is an observational study because I'm comparing the preferences of steak between men and women.

```python
import pandas as pd
import seaborn as sns
```

```python
steak = pd.read_csv("steak.csv")
steak
```

```python
steak_type = steak["How do you like your steak prepared?"]
gender = steak["Gender"]
steak_type
```

```python
gender
```

```python
plot_df = pd.DataFrame({"steak_type": steak_type, "gender": gender}).dropna()
plot_df = plot_df[plot_df["steak_type"] != "Response"]

order = ["Rare", "Medium rare", "Medium", "Medium Well", "Well"]

g = sns.catplot(
    data=plot_df,
    x="steak_type",
    hue="gender",
    kind="count",
    order=order,
    height=5,
    aspect=1.7,
)
g.set_axis_labels("How do you like your steak prepared?", "Number of respondents")
g.figure.suptitle("Steak Preparation Preferences by Gender", y=1.03)
g.legend.set_title("Gender")
```

The men seam to like medium rare steak while the women seam to like their steaks more well done.

The bar chart shows that men like medium to medium rare steaks and the women like medium well to well done steaks.
