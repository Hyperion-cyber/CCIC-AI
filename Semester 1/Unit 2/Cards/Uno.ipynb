### My Question
What's the joint probability that I pull a green reverse card from a deck of uno cards?

```python
import random
import seaborn as sns
```

```python
green_reverse_count = 0
```

```python
trials = []
probabilities = []
```

```python
probability = green_reverse_count / 1000
```

```python
probability
```

```python
for i in range(1000):
    card = random.randint(0, 107)
    if card == 0 or card == 1:
        green_reverse_count += 1
    if i + 1 in [100, 250, 500, 750, 1000]:
        probability = green_reverse_count / (i + 1)
        trials.append(i + 1)
        probabilities.append(probability)
```

```python
sns.lineplot(x=trials, y=probabilities)
```

```python
print(len(trials))
print(len(probabilities))
```
