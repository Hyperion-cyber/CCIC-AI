### My Question
What's the joint probability that I pull a green reverse card from a deck of uno cards?

```python
import random
```

```python
green_reverse_count = 0
```

```python
for i in range(1000):
    card = random.randint(0, 107)
    if card == 0 or card == 1:
        green_reverse_count += 1
```

```python
probability = green_reverse_count / 1000
```
