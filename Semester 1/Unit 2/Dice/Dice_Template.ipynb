```python
import pandas as pd
import numpy.random as rand
from fractions import Fraction
```

```python
sample_space = [1,2,3,4]
print(f"Number of possible outcomes:  {len(sample_space)}")
```

```python
ideal_probability = Fraction(1/len(sample_space)).limit_denominator()
print(f"ideal probability: {ideal_probability}")
```

```python
rolls = rand.randint(low =1, high=7, size =1000)
rolls_series = pd.Series(data=rolls)
```

```python
counts = rolls_series.value_counts()
```
